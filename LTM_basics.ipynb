```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
namespace1 = ('users','u1')
```

### Creating Memories

```python
store.put(namespace1, "1", {"data": "User likes pasta"})
store.put(namespace1, "2", {"data": "User prefers grid style navigation"})
```

```python
# another namespace
namespace2 = ("user", "u2")

# Add memories
store.put(namespace2, "1", {"data": "User likes pizza"})
store.put(namespace2, "2", {"data": "User prefers list style navigation"})
```

### Retrieving Memories

```python
store.get(namespace1, "1")
```

```python
store.get(namespace1, "2")
```

```python
items = store.search(namespace2)

for item in items:
    print(item.value)
```
